## train/validation/test split 구성

1. source별 최종 데이터 개수 확인
2. source 비율을 유지하면서 train/val/test split
3. patient/file 단위 중복 방지
4. split CSV 저장
5. Dataset 클래스에서 split CSV를 읽도록 구성

- train: 70%
- validation: 15%
- test: 15%

```python
import pandas as pd
from pathlib import Path

WORK_DIR = Path(r"C:\Users\comds\Desktop\졸업작품\vscode_script")

df_t11_t12_final = pd.read_pickle(
    WORK_DIR / "df_t11_t12_final_candidates.pkl"
)
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd

df = df_t11_t12_final.copy()

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["source"],
    random_state=42,
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["source"],
    random_state=42,
)

train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df["split"] = "train"
val_df["split"] = "val"
test_df["split"] = "test"

df_split = pd.concat([train_df, val_df, test_df], ignore_index=True)

display(pd.crosstab(df_split["split"], df_split["source"]))
display(pd.crosstab(df_split["split"], df_split["source"], normalize="index"))

df_split.to_csv(
    "df_t11_t12_train_val_test_split.csv",
    index=False,
    encoding="utf-8-sig",
)

df_split.to_pickle("df_t11_t12_train_val_test_split.pkl")
```
